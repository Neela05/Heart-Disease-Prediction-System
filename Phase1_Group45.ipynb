```python
# Renmaing the column name for good fuctionality
new_column_names = {
    'ChestPainType': 'pain_type',
    'RestingBP': 'resting_bp',
    'FastingBS': 'fasting_bs',
    'RestingECG': 'resting_ecg',
    'ExerciseAngina': 'ex_angina',
    'MaxHR': 'max_heart_rate',
    'Oldpeak': 'old_peak',
    'ST_Slope': 'st_slope',
    'HeartDisease': 'heart_disease'
}

# Rename columns and strip whitespace and convert to lowercase
df_new = df.rename(columns=new_column_names).rename(columns=str.strip).rename(columns=str.lower).copy()
```

```python
df_new.tail()
```

```python
# To see the different values and it count
df_new['pain_type'].value_counts()
```

```python
# Check and drop the zero values in specific columns
columns_to_check = ['resting_bp', 'cholesterol','max_heart_rate','age']


if (df_new[columns_to_check] == 0).any().any():

    df_new = df_new[~(df_new[columns_to_check] == 0).any(axis=1)]
    print("Rows with 0 values in columns {} dropped.".format(columns_to_check))
else:
    print("Columns {} do not contain 0 values.".format(columns_to_check))


print("DataFrame after dropping rows with 0 values in columns {}:".format(columns_to_check))

df_new
```

## Data Exploration & Visualisation <a id='dataexploration'></a>

With our dataset now deemed clean, we're prepared to commence visualizing and exploring each of its features.

### Univariate Visualisation

##### Bar plot of peak of ST slope segment

The figure 1.1 is countplot which shows the frequency of each ST slope type using different colors for better visualization.As we can see from the figure 1.1  Flat slope is the max count in the dataset.

```python
# The barplot showing the count of different ST slope
custom_colors = ["skyblue", "salmon", "lightgreen"]
sns.countplot(x=df_new['st_slope'], palette=custom_colors)
plt.title('Figure 1.1: Distribution of Slope of peak ST segment ', fontweight='bold')
plt.ylim(0, df_new['st_slope'].value_counts().max() * 1.1)
plt.show()
```

##### Histrogram of Age

The figure 1.2 is the histogram illustrating the distribution of ages within the dataset. Each bar represents the frequency (count) of individuals falling within a specific age range.

```python
# Ploting the Histrogram for Age distribution
plt.figure(figsize=(20, 14))
plt.subplot(3, 2, 1)
sns.histplot(df_new['age'], bins=20, kde=True, color='skyblue', line_kws={'color': 'orange'})
plt.title('Figure 1.2: Histogram of Age Distribution', fontweight='bold')
plt.xlabel('Age', fontweight='bold')
max_count = sns.histplot(df_new['age'], bins=20, kde=True).get_ylim()[1]
plt.ylim(0, max_count * 1.1)
plt.show()

```

##### Bar plot of Chest pain Type

The below figure 1.3 is a bar plot illustrating the frequency of different types of chest pains within the dataset. Each bar represents the count of occurrences for a specific type of chest pain.

```python
# Plotting the bar plot fo the frequency of different chest pain types
plt.figure(figsize=(14, 6))
fig = sns.countplot(x='pain_type', data=df_new, palette='viridis', order=df_new['pain_type'].value_counts().index)
fig.set_title('Figure 1.3: Type of Chest Pains', fontsize=16, fontweight='bold')
fig.set_xlabel('Pain Type', fontsize=14, fontweight='bold')
plt.ylim(0, df_new['pain_type'].value_counts().max() * 1.1)
plt.show()
```

##### Pie chart of Sex

Figure 1.4 is a pie chart representing the distribution of sex (gender). Each segment of the pie chart represents a sex category, with the size of each segment proportional to the percentage of individuals belonging to that category in the dataset.

```python
# Plotting the pie plot to observe total count of sex

plt.rcParams.update({'font.size': 15})
ax = df_new['sex'].value_counts().plot.pie(explode=[0.1, 0.1], autopct='%1.2f%%', shadow=True, colors=['skyblue', 'lightcoral'])
ax.set_title(label="Figure 1.4: Pie chart of the Sex Distibution", fontsize=20, color='black',fontweight='bold')
plt.legend(labels=['Male', 'Female'])
plt.axis('off')
plt.show()
```

### Two-Variable Visualisation

##### Regression plot between Cholestrol and Age

Figure 2.1 is a regression plot illustrating the relationship between cholesterol levels and age. Each data point represents an individual's cholesterol level and age. The plot includes a regression line, which shows the overall trend in the data.

```python
# Plotting the regression plot to see the trend between cholestrol level and age

sns.set_style("whitegrid")
sns.regplot(x=df_new['cholesterol'], y=df_new['age'], scatter_kws={'color': 'skyblue', 'alpha': 0.5, 's': 100},
            line_kws={'color': 'salmon', 'linewidth': 2})

plt.xlabel('Cholesterol', fontsize=14,fontweight='bold')
plt.ylabel('Age', fontsize=14,fontweight='bold')
plt.title('Figure 2:1:Regression Plot of Cholesterol vs Age', fontsize=16,fontweight='bold')
plt.show()
```

##### Scatter plot of Cholestrol by Sex

Figure 2.2: Boxplot of Cholesterol by Gender  represents the distribution of cholesterol levels among different genders. This plot helps visualize the spread and central tendency of cholesterol levels across different sexes, allowing for  comparison between the two groups.

```python
# Plotting the Boxplot for the sex and cholestrol
plt.figure(figsize=(12, 8))
sns.boxplot(x=df_new['sex'], y=df_new['cholesterol'], palette='Set2')
plt.xlabel('Sex', fontsize=14,fontweight='bold')
plt.ylabel('Cholesterol', fontsize=14,fontweight='bold')
plt.title('Figure 2.2: Boxplot of Cholesterol by Gender', fontsize=16,fontweight='bold')
plt.show()
```

##### Bar chart for Pain type by Sex

The plot depicts a bar chart illustrating the distribution of different types of chest pain among different genders within the dataset.

```python
# Plotting the bar chart for different chest pain of different type by gender
plt.figure(figsize=(14, 6))
sns.countplot(x='sex', hue='pain_type', data=df_new, palette='viridis', order=df_new['sex'].value_counts().index)
plt.title('Figure 2.3: Distribution of Pain Types by Gender', fontsize=16,fontweight='bold')
plt.xlabel('Sex', fontsize=14, fontweight='bold')
plt.ylabel('Count', fontsize=14, fontweight='bold')
plt.legend(title='Pain Type', title_fontsize='14', fontsize='12')
plt.show()
```

##### Scatter plot with hsitrogram for Age and Max Heart Rate

The jointplot includes scatter plots for each feature( Age and Max Heart Rate), histograms for each variable's distribution, and a regression line representing the relationship between the two variables.

```python
# Plotting the jointplot to explore the relationship between age and maximum heart rate
sns.set_style("whitegrid")

jointplot = sns.jointplot(data=df_new, x="age", y="max_heart_rate", kind="reg", scatter_kws={'alpha':0.5})

jointplot.set_axis_labels('Age', 'Max heart rate', fontsize=14, fontweight='bold')

plt.suptitle('Figure 2.4: Relationship between Age and Max Heart Rate', fontsize=16, fontweight='bold')

plt.tight_layout()

plt.show()

```

### Three-Variable Visualisation

##### Heat map for Sex,Pain Type and Heart Disease Occurrence  

The distribution of heart disease occurrences based on the combination of sex (male or female) and pain type. Each cell in the heatmap represents the count of individuals with a specific combination of sex, pain type, and heart disease status.

```python
# Plotting the heat map for Sex,Pain type and Heart disease
heatmap_data = df_new.pivot_table(index='sex', columns=['pain_type', 'heart_disease'], aggfunc='size', fill_value=0)
plt.figure(figsize=(10, 8))
sns.heatmap(heatmap_data, annot=True, cmap='viridis', fmt='d')
plt.title('Figure 3.1: Heatmap of Sex, Pain Type and Heart Disease Occurrence', fontsize=16,fontweight='bold')
plt.xlabel('Pain Type, Heart Disease', fontsize=14,fontweight='bold')
plt.ylabel('Sex', fontsize=14,fontweight='bold')
plt.show()
```

##### Scatter Plot of Age, Cholestrol and Sex

The scatter plot depicts the relationship between age and cholesterol levels, with gender distinguished by color. plot helps to visualize patterns or trends in age and cholesterol levels vary across genders.

```python
# Plotting a scatter plot of 'Age' and 'Cholesterol' and 'Sex'
plt.figure(figsize=(10, 6))
scatterplot = sns.scatterplot(x='age', y='cholesterol', hue='sex', data=df_new, palette='Set1')
plt.title('figure 3.2: Scatter Plot of Relationship Between Age and Cholesterol Levels by Gender', fontsize=16,
          fontweight='bold')
plt.xlabel('Age', fontsize=14, fontweight='bold')
plt.ylabel('Cholesterol', fontsize=14, fontweight='bold')
legend = scatterplot.legend(title='Sex')
legend.texts[0].set_text('M')
legend.texts[1].set_text('F')
for text in legend.texts:
    text.set_fontweight('bold')
plt.show()
```

##### Kernel Density Estimation (KDE) Plot of Age,Resting Blood Pressure and Old Peak

KDE plot depicts the probability density function of the age, resting blood pressure, and old peak. It gives an estimation of the probability density of each variable, showing the distribution of values and indicating areas of higher or lower density within each variable's range.

```python
#p Plotting the KDE for the Age , Resting Blood Pressure and Old Peak
sns.kdeplot(data=df_new[['age', 'resting_bp', 'old_peak']], fill=True)
plt.xlabel('Range of density estimates', fontsize=14,fontweight='bold')
plt.ylabel('Density of the data', fontsize=14,fontweight='bold')
plt.legend(['Age', 'Resting Blood Pressure', 'Old Peak'], fontsize=12)
plt.title('Figure 3.3: Distribution of Age, Resting Blood Pressure, and Old Peak: Kernel Density Estimation (KDE) Plot.',
          fontsize=16,fontweight='bold')
plt.show()
```

##### Group Bar chart of Max Heart Rate,Type of Pain and Sex

This grouped bar plot depicts the average maximum heart rate (MaxHR) categorized by sex and pain type. Each bar represents the average MaxHR for a specific combination of sex and pain type.

```python
# Plotting the bar chart of different types of pain with Max heart rate by sex

plt.figure(figsize=(10, 6))
sns.barplot(x='sex', y='max_heart_rate', hue='pain_type', data=df_new, palette='viridis')
plt.title('Figure 3.4: Grouped Bar Plot of Maximum Heart Rate by Sex and Pain Type', fontsize=16, fontweight='bold')
plt.xlabel('Sex', fontsize=14, fontweight='bold')
plt.ylabel('Max Heart Rate', fontsize=14, fontweight='bold')
plt.legend(title='Type of Pain')
plt.xticks(ticks=[0, 1], labels=['Male', 'Female'], fontweight='bold')
plt.show()

```

##### Scatter plot of Age,Resting Blood Pressure By Heart Disease

The scatter plot shows the relationship between age and resting blood pressure, with the data points  to represent the presence or absence of heart disease.This hepls to visualize potential correlations between age, resting blood pressure, and the likelihood of heart disease.


```python
# Plotting the Scatter plot for the relationship between age, resting blood pressure and heart disease
plt.figure(figsize=(10, 6))
scatterplot = sns.scatterplot(x='age', y='resting_bp', hue='heart_disease', data=df_new, palette='Set2')
plt.title('Figure 3.5: Scatter Plot of Relationship Between Age and Resting Blood Pressure by Heart Disease',
          fontsize=16, fontweight='bold')
plt.xlabel('Age', fontsize=14, fontweight='bold')
plt.ylabel('Resting Blood Pressure', fontsize=14, fontweight='bold')

legend = scatterplot.legend(title='Heart Disease')
legend.texts[0].set_text('No')
legend.texts[1].set_text('Yes')
for text in legend.texts:
    text.set_fontweight('bold')
plt.show()
```

## Literature Review

#### Literature Review: Heart Disease Prediction System using Machine Learning

Several investigations have delved into employing machine learning for heart disease prediction. Shuge Ouyang (2022)[1] and Chua, S.(2022)[2] each devised systems utilizing various algorithms; Shuge Ouyang concentrated on Support Vector Machine and decision trees, while Chua, S. explored Naïve Bayes, K-Nearest Neighbor, Support Vector Machine, Random Forest, and Decision Tree. Shah, W.(2021)[3] established a dependable prediction system using the Random Forest algorithm, whereas Yadav (2021)[4] achieved a remarkable 96% accuracy through the application of a genetic algorithm for feature optimization in the Naïve Bayes model. Collectively, these studies underscore the potential of machine learning in accurately forecasting heart disease.

#### 1. Introduction to Heart Disease Prediction:
This definition from the Heart disease encompasses a range of conditions affecting the heart and blood vessels, including coronary artery disease, heart failure, and arrhythmias. Traditional risk assessment methods rely on clinical factors such as age, gender, blood pressure, cholesterol levels, and smoking status. However, machine learning offers the potential to integrate diverse data sources and uncover complex patterns that may elude conventional approaches(Subramani,S.2023)[5].

#### 2. Data Sources and Feature Selection:
An essential step in creating predictive models for heart disease is choosing suitable features. Researchers have utilized varied datasets, incorporating electronic health records, medical images, genetic data, and lifestyle elements. Techniques like recursive feature elimination, principal component analysis, and information gain have been used to pinpoint the most informative variables for training models(Arin,S.2024)[6].

#### 3. Machine Learning Algorithms:
Several studies have investigated the application of machine learning algorithms in heart disease prediction. H. Yu (2023)[7] proposed a system that integrated optimal attributes and algorithms to enhance diagnosis and alleviate hospital workload. Kanimozhi, S.(2024)[8] devised a system utilizing KNN and decision tree algorithms, offering advanced prediction capabilities and improved accuracy. Jeyaraman, Gowri.(2022)[9] established a dependable prediction system employing the Random Forest algorithm, achieving notable performance and accuracy levels. Tota, M.(2022)[10] underscored the promise of machine learning in heart disease prediction by analyzing patient data and identifying key risk factors. Together, these studies underscore the potential of machine learning algorithms in enhancing the precision and efficacy of heart disease prediction systems.

#### 4. Performance Evaluation Metrics:
Several investigations have delved into utilizing machine learning algorithms for heart disease prediction, each employing diverse methodologies and assessment criteria. Gabriel, J.(2024)[11] created a web application utilizing commonly used machine learning algorithms, whereas Singh, M. S.(2020)[12] assessed classification algorithms, identifying Random Forest as exhibiting the highest accuracy. Together, these studies underscore the potential of machine learning in heart disease prediction, with an emphasis on accuracy, sensitivity, specificity, and other performance evaluation metrics.

#### 5. Challenges and Limitations:
Developing systems to predict heart disease is a major focus of research. Scientists are using patient records and various computer programs to improve the accuracy of these predictions. However, there are challenges with these systems. L. P. Koyi (2021)[13] discusses issues like how to describe the information, identifying risk factors, and establishing connections between them. Rahman (2022)[14] emphasizes the importance of fast and accurate predictions. They used computer programs to achieve high accuracy and explored different ways to organize information. Despite progress, there are still problems with the quality of information, understanding it, and making it beneficial for everyone. More research is necessary to address these challenges.

#### 6. Recent Advances and Future Directions:
Recent Studies has been investigating inventive methods to improve the precision and dependability of models predicting heart disease. This involves integrating diverse data sources like wearable tech, genetic data, and factors affecting health. Advanced machine learning methods, especially convolutional and recurrent neural networks, have demonstrated potential in extracting important patterns from raw data and enhancing predictive accuracy. In a subsequent study conducted by Prof. S. B. Pagrut (2023)[15], advancements were made in employing machine learning methodologies to categorize patients according to diverse medical characteristics, resulting in superior accuracy compared to earlier classification models.

#### 7. Conclusion:
In summary, machine learning-driven heart disease prediction systems offer significant potential for enhancing risk assessment and preventive healthcare. Through the utilization of varied data streams and sophisticated analysis, these models can deepen our comprehension of cardiovascular risk factors, enabling individuals to make informed decisions about their lifestyles.
